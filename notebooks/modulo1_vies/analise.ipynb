```python
duplicatas_linhas = df.duplicated().sum()

print(f"Linhas completamente duplicadas: {duplicatas_linhas}")
```

```python
# Duplicidade pelo nome do planeta
if "pl_name" in df.columns:
    duplicatas_planeta = df["pl_name"].duplicated().sum()
    print(f"Nomes de planeta repetidos: {duplicatas_planeta}")
```

```python
colunas_principais = [
    "pl_name",
    "hostname",
    "pl_orbper",
    "pl_orbsmax",
    "pl_rade",
    "pl_bmasse",
    "pl_eqt",
    "st_teff",
    "sy_dist",
    "discoverymethod",
    "disc_year",
    "disc_facility"
]

# Mantém apenas as colunas que realmente existem na versão do CSV
colunas_disponiveis = [c for c in colunas_principais if c in df.columns]

df_eda = df[colunas_disponiveis].copy()

print(f"Variáveis selecionadas: {len(colunas_disponiveis)}")
print(colunas_disponiveis)
```

```python
df_eda = df_eda.drop_duplicates(subset="pl_name").copy()

colunas_numericas = [
    "pl_orbper",
    "pl_orbsmax",
    "pl_rade",
    "pl_bmasse",
    "pl_eqt",
    "st_teff",
    "sy_dist",
    "disc_year"
]

for coluna in colunas_numericas:
    if coluna in df_eda.columns:
        df_eda[coluna] = pd.to_numeric(df_eda[coluna], errors="coerce")

print(f"Registros após remoção de duplicatas: {len(df_eda):,}")
```

```python
# Remove apenas valores impossíveis nas grandezas básicas
condicao = pd.Series(True, index=df_eda.index)

for coluna in ["pl_orbper", "pl_orbsmax", "pl_rade", "pl_bmasse", "pl_eqt", "st_teff", "sy_dist"]:
    if coluna in df_eda.columns:
        condicao &= df_eda[coluna].isna() | (df_eda[coluna] > 0)

df_eda = df_eda[condicao].copy()

print(f"Registros após limpeza básica: {len(df_eda):,}")
```

```python
variaveis_fisicas = [
    "pl_orbper",
    "pl_orbsmax",
    "pl_rade",
    "pl_bmasse",
    "pl_eqt",
    "st_teff",
    "sy_dist"
]

variaveis_fisicas = [c for c in variaveis_fisicas if c in df_eda.columns]

estatisticas = df_eda[variaveis_fisicas].describe().T
estatisticas
```

```python
resumo = pd.DataFrame({
    "média": df_eda[variaveis_fisicas].mean(),
    "mediana": df_eda[variaveis_fisicas].median(),
    "desvio_padrão": df_eda[variaveis_fisicas].std(),
    "mínimo": df_eda[variaveis_fisicas].min(),
    "máximo": df_eda[variaveis_fisicas].max()
})

resumo
```

```python
for coluna in variaveis_fisicas:
    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=df_eda,
        x=coluna,
        bins=40,
        kde=True
    )

    plt.title(f"Distribuição de {coluna}")
    plt.xlabel(coluna)
    plt.ylabel("Quantidade")

    plt.tight_layout()
    plt.show()
```

```python
variaveis_log = [
    c for c in ["pl_rade", "pl_bmasse", "pl_orbper", "pl_orbsmax", "sy_dist"]
    if c in df_eda.columns
]

for coluna in variaveis_log:
    dados = df_eda[coluna].dropna()
    dados = dados[dados > 0]

    plt.figure(figsize=(8, 4))

    sns.histplot(
        x=np.log10(dados),
        bins=40,
        kde=True
    )

    plt.title(f"Distribuição de log10({coluna})")
    plt.xlabel(f"log10({coluna})")
    plt.ylabel("Quantidade")

    plt.tight_layout()
    plt.show()
```

```python
for coluna in variaveis_fisicas:
    plt.figure(figsize=(9, 2.5))

    sns.boxplot(x=df_eda[coluna].dropna())

    plt.title(f"Boxplot de {coluna}")
    plt.xlabel(coluna)

    plt.tight_layout()
    plt.show()
```

Correlação entre as variáveis físicas

```python
# Selecionando as variáveis físicas
variaveis = [
    "pl_rade",
    "pl_bmasse",
    "pl_orbper",
    "pl_orbsmax",
    "pl_eqt",
    "st_teff",
    "sy_dist"
]

corr = df_eda[variaveis].corr(method="spearman")

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Matriz de correlação de Spearman")
plt.tight_layout()
plt.show()
```
